```python
import os
import json
import yaml
import joblib
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

warnings.filterwarnings("ignore")

try:
    import tensorflow as tf
    from tensorflow.keras.models import Sequential
    from tensorflow.keras.layers import Dense, Dropout
except Exception:
    tf = None


BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Slum Risk Prediction System"
CSV_PATH = os.path.join(BASE_DIR, "D25-HousingandSlumPopulation_8_2.csv")

os.makedirs(BASE_DIR, exist_ok=True)

PKL_PATH = os.path.join(BASE_DIR, "slum_risk_model.pkl")
H5_PATH = os.path.join(BASE_DIR, "slum_risk_model.h5")
YAML_PATH = os.path.join(BASE_DIR, "slum_risk_config.yaml")
JSON_PATH = os.path.join(BASE_DIR, "slum_risk_metrics.json")

RESULT_CSV_PATH = os.path.join(BASE_DIR, "slum_risk_results.csv")
PREDICTION_CSV_PATH = os.path.join(BASE_DIR, "slum_risk_predictions.csv")
MODEL_COMPARISON_CSV = os.path.join(BASE_DIR, "model_comparison_results.csv")

ACCURACY_GRAPH_PATH = os.path.join(BASE_DIR, "accuracy_graph.png")
COMPARISON_GRAPH_PATH = os.path.join(BASE_DIR, "model_comparison_graph.png")
HEATMAP_PATH = os.path.join(BASE_DIR, "confusion_matrix_heatmap.png")
RESULT_GRAPH_PATH = os.path.join(BASE_DIR, "result_distribution_graph.png")
PREDICTION_GRAPH_PATH = os.path.join(BASE_DIR, "prediction_distribution_graph.png")
FEATURE_IMPORTANCE_GRAPH_PATH = os.path.join(BASE_DIR, "feature_importance_graph.png")


def clean_numeric(value):
    if pd.isna(value):
        return np.nan

    value = str(value)
    value = value.replace(",", "")
    value = value.replace("%", "")
    value = value.strip()

    try:
        return float(value)
    except:
        return np.nan


def prepare_dataset(df):
    required_numeric_cols = [
        "Ward No.",
        "No. Notified Slums",
        "No. of Recognised Slums",
        "No. of Identfied Slums",
        "Slum Population - Total",
        "Slum Population - Male",
        "Slum Population - Female",
        "Slum Population - Child (0-6)",
        "Population - SC Category",
        "Population - ST Category",
        "Literacy Rate (%)",
    ]

    for col in required_numeric_cols:
        if col in df.columns:
            df[col] = df[col].apply(clean_numeric)
        else:
            df[col] = 0

    df[required_numeric_cols] = df[required_numeric_cols].fillna(0)

    df["Total Slums"] = (
        df["No. Notified Slums"]
        + df["No. of Recognised Slums"]
        + df["No. of Identfied Slums"]
    )

    df["Slum Density Score"] = (
        df["Slum Population - Total"] / (df["Total Slums"] + 1)
    )

    df["Gender Gap"] = abs(
        df["Slum Population - Male"] - df["Slum Population - Female"]
    )

    df["Child Population Ratio"] = (
        df["Slum Population - Child (0-6)"] /
        (df["Slum Population - Total"] + 1)
    )

    df["SC ST Population"] = (
        df["Population - SC Category"] + df["Population - ST Category"]
    )

    df["Vulnerability Score"] = (
        df["Slum Population - Total"] * 0.40
        + df["SC ST Population"] * 0.20
        + df["Slum Population - Child (0-6)"] * 0.20
        + df["Slum Density Score"] * 0.15
        - df["Literacy Rate (%)"] * 0.05
    )

    q1 = df["Vulnerability Score"].quantile(0.33)
    q2 = df["Vulnerability Score"].quantile(0.66)

    def risk_label(score):
        if score <= q1:
            return "Low"
        elif score <= q2:
            return "Medium"
        else:
            return "High"

    df["Risk Level"] = df["Vulnerability Score"].apply(risk_label)

    return df


def save_bar_graph(data, x_col, y_col, title, xlabel, ylabel, path):
    plt.figure(figsize=(10, 6))
    sns.barplot(data=data, x=x_col, y=y_col)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.xticks(rotation=30)
    plt.tight_layout()
    plt.savefig(path, dpi=300)
    plt.show()


def save_count_graph(df, column, title, xlabel, ylabel, path):
    plt.figure(figsize=(8, 5))
    sns.countplot(data=df, x=column, order=df[column].value_counts().index)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.tight_layout()
    plt.savefig(path, dpi=300)
    plt.show()


def main():
    print("Loading dataset...")
    df = pd.read_csv(CSV_PATH)

    print("Original dataset shape:", df.shape)

    df = prepare_dataset(df)

    feature_cols = [
        "No. Notified Slums",
        "No. of Recognised Slums",
        "No. of Identfied Slums",
        "Slum Population - Total",
        "Slum Population - Male",
        "Slum Population - Female",
        "Slum Population - Child (0-6)",
        "Population - SC Category",
        "Population - ST Category",
        "Literacy Rate (%)",
        "Total Slums",
        "Slum Density Score",
        "Gender Gap",
        "Child Population Ratio",
        "SC ST Population",
        "Vulnerability Score",
    ]

    X = df[feature_cols]
    y = df["Risk Level"]

    label_encoder = LabelEncoder()
    y_encoded = label_encoder.fit_transform(y)

    if len(df) < 10 or len(set(y_encoded)) < 2:
        print("\nDataset is too small for proper ML training.")
        print("Saving rule-based outputs only.\n")

        df["Predicted Risk Level"] = df["Risk Level"]
        df["Prediction Status"] = "Rule-Based"

        df.to_csv(RESULT_CSV_PATH, index=False)
        df.to_csv(PREDICTION_CSV_PATH, index=False)

        save_count_graph(
            df,
            "Risk Level",
            "Slum Risk Level Distribution",
            "Risk Level",
            "Count",
            RESULT_GRAPH_PATH
        )

        save_count_graph(
            df,
            "Predicted Risk Level",
            "Predicted Slum Risk Distribution",
            "Predicted Risk Level",
            "Count",
            PREDICTION_GRAPH_PATH
        )

        config = {
            "project_name": "Slum Risk Prediction System",
            "status": "rule_based_only",
            "reason": "Dataset too small for proper model training",
            "dataset_path": CSV_PATH,
            "feature_columns": feature_cols,
            "target_column": "Risk Level",
            "outputs": {
                "result_csv": RESULT_CSV_PATH,
                "prediction_csv": PREDICTION_CSV_PATH,
                "result_graph": RESULT_GRAPH_PATH,
                "prediction_graph": PREDICTION_GRAPH_PATH
            }
        }

        metrics = {
            "status": "rule_based_only",
            "rows": int(len(df)),
            "risk_classes": list(label_encoder.classes_),
            "message": "Add more rows for ML accuracy, comparison graph, and heatmap."
        }

        joblib.dump(
            {
                "model_type": "Rule-Based",
                "feature_columns": feature_cols,
                "label_encoder": label_encoder
            },
            PKL_PATH
        )

        with open(YAML_PATH, "w") as f:
            yaml.dump(config, f, sort_keys=False)

        with open(JSON_PATH, "w") as f:
            json.dump(metrics, f, indent=4)

        print("Saved rule-based files successfully.")
        return

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y_encoded,
        test_size=0.2,
        random_state=42,
        stratify=y_encoded
    )

    models = {
        "Logistic Regression": Pipeline([
            ("scaler", StandardScaler()),
            ("classifier", LogisticRegression(max_iter=1000))
        ]),
        "Decision Tree": DecisionTreeClassifier(random_state=42),
        "Random Forest": RandomForestClassifier(
            n_estimators=200,
            random_state=42,
            class_weight="balanced"
        ),
        "Gradient Boosting": GradientBoostingClassifier(random_state=42)
    }

    comparison_results = []
    trained_models = {}

    print("Training models...")

    for model_name, model in models.items():
        model.fit(X_train, y_train)
        preds = model.predict(X_test)
        acc = accuracy_score(y_test, preds)

        comparison_results.append({
            "Model": model_name,
            "Accuracy": acc
        })

        trained_models[model_name] = model
        print(f"{model_name} Accuracy: {acc:.4f}")

    comparison_df = pd.DataFrame(comparison_results)
    comparison_df.to_csv(MODEL_COMPARISON_CSV, index=False)

    best_row = comparison_df.sort_values(by="Accuracy", ascending=False).iloc[0]
    best_model_name = best_row["Model"]
    best_model = trained_models[best_model_name]

    print("\nBest Model:", best_model_name)

    y_pred = best_model.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)

    df["Predicted Risk Encoded"] = best_model.predict(X)
    df["Predicted Risk Level"] = label_encoder.inverse_transform(
        df["Predicted Risk Encoded"]
    )

    df["Prediction Status"] = np.where(
        df["Risk Level"] == df["Predicted Risk Level"],
        "Correct",
        "Incorrect"
    )

    df.to_csv(RESULT_CSV_PATH, index=False)

    prediction_df = df[
        [
            "Risk Level",
            "Predicted Risk Level",
            "Prediction Status",
            "Vulnerability Score",
            "Slum Density Score",
            "Total Slums"
        ]
    ]

    prediction_df.to_csv(PREDICTION_CSV_PATH, index=False)

    model_bundle = {
        "model_type": best_model_name,
        "model": best_model,
        "feature_columns": feature_cols,
        "label_encoder": label_encoder,
        "risk_classes": list(label_encoder.classes_)
    }

    joblib.dump(model_bundle, PKL_PATH)

    metrics = {
        "status": "trained",
        "best_model": best_model_name,
        "accuracy": float(accuracy),
        "classification_report": classification_report(
            y_test,
            y_pred,
            target_names=label_encoder.classes_,
            output_dict=True
        ),
        "confusion_matrix": confusion_matrix(y_test, y_pred).tolist(),
        "model_comparison": comparison_results,
        "output_files": {
            "pkl": PKL_PATH,
            "h5": H5_PATH,
            "yaml": YAML_PATH,
            "json": JSON_PATH,
            "result_csv": RESULT_CSV_PATH,
            "prediction_csv": PREDICTION_CSV_PATH,
            "accuracy_graph": ACCURACY_GRAPH_PATH,
            "comparison_graph": COMPARISON_GRAPH_PATH,
            "heatmap": HEATMAP_PATH,
            "result_graph": RESULT_GRAPH_PATH,
            "prediction_graph": PREDICTION_GRAPH_PATH,
            "feature_importance_graph": FEATURE_IMPORTANCE_GRAPH_PATH
        }
    }

    with open(JSON_PATH, "w") as f:
        json.dump(metrics, f, indent=4)

    config = {
        "project_name": "Slum Risk Prediction System",
        "dataset_path": CSV_PATH,
        "base_directory": BASE_DIR,
        "target_column": "Risk Level",
        "best_model": best_model_name,
        "feature_columns": feature_cols,
        "risk_classes": list(label_encoder.classes_),
        "artifacts": metrics["output_files"]
    }

    with open(YAML_PATH, "w") as f:
        yaml.dump(config, f, sort_keys=False)

    save_bar_graph(
        comparison_df,
        "Model",
        "Accuracy",
        "Model Accuracy Comparison",
        "Model",
        "Accuracy",
        ACCURACY_GRAPH_PATH
    )

    save_bar_graph(
        comparison_df,
        "Model",
        "Accuracy",
        "ML Model Comparison Graph",
        "Model",
        "Accuracy Score",
        COMPARISON_GRAPH_PATH
    )

    cm = confusion_matrix(y_test, y_pred)

    plt.figure(figsize=(8, 6))
    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=label_encoder.classes_,
        yticklabels=label_encoder.classes_
    )
    plt.title("Confusion Matrix Heatmap")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.tight_layout()
    plt.savefig(HEATMAP_PATH, dpi=300)
    plt.show()

    save_count_graph(
        df,
        "Risk Level",
        "Actual Slum Risk Result Distribution",
        "Actual Risk Level",
        "Count",
        RESULT_GRAPH_PATH
    )

    save_count_graph(
        df,
        "Predicted Risk Level",
        "Predicted Slum Risk Distribution",
        "Predicted Risk Level",
        "Count",
        PREDICTION_GRAPH_PATH
    )

    if best_model_name == "Random Forest":
        importances = best_model.feature_importances_
        importance_df = pd.DataFrame({
            "Feature": feature_cols,
            "Importance": importances
        }).sort_values(by="Importance", ascending=False)

        plt.figure(figsize=(12, 7))
        sns.barplot(data=importance_df, x="Importance", y="Feature")
        plt.title("Feature Importance Graph")
        plt.xlabel("Importance")
        plt.ylabel("Feature")
        plt.tight_layout()
        plt.savefig(FEATURE_IMPORTANCE_GRAPH_PATH, dpi=300)
        plt.show()

    if tf is not None:
        scaler = StandardScaler()
        X_train_scaled = scaler.fit_transform(X_train)
        X_test_scaled = scaler.transform(X_test)

        nn_model = Sequential([
            Dense(64, activation="relu", input_shape=(X_train_scaled.shape[1],)),
            Dropout(0.2),
            Dense(32, activation="relu"),
            Dropout(0.2),
            Dense(len(label_encoder.classes_), activation="softmax")
        ])

        nn_model.compile(
            optimizer="adam",
            loss="sparse_categorical_crossentropy",
            metrics=["accuracy"]
        )

        history = nn_model.fit(
            X_train_scaled,
            y_train,
            validation_data=(X_test_scaled, y_test),
            epochs=50,
            batch_size=8,
            verbose=1
        )

        nn_model.save(H5_PATH)

        history_df = pd.DataFrame(history.history)
        history_csv_path = os.path.join(BASE_DIR, "h5_training_history.csv")
        history_graph_path = os.path.join(BASE_DIR, "h5_training_accuracy_graph.png")

        history_df.to_csv(history_csv_path, index=False)

        plt.figure(figsize=(10, 6))
        plt.plot(history.history["accuracy"], label="Training Accuracy")
        plt.plot(history.history["val_accuracy"], label="Validation Accuracy")
        plt.title("H5 Neural Network Accuracy Graph")
        plt.xlabel("Epochs")
        plt.ylabel("Accuracy")
        plt.legend()
        plt.tight_layout()
        plt.savefig(history_graph_path, dpi=300)
        plt.show()

    print("\nAll files saved successfully in:")
    print(BASE_DIR)

    print("\nSaved files:")
    print("PKL:", PKL_PATH)
    print("H5:", H5_PATH)
    print("YAML:", YAML_PATH)
    print("JSON:", JSON_PATH)
    print("Result CSV:", RESULT_CSV_PATH)
    print("Prediction CSV:", PREDICTION_CSV_PATH)
    print("Model Comparison CSV:", MODEL_COMPARISON_CSV)
    print("Accuracy Graph:", ACCURACY_GRAPH_PATH)
    print("Comparison Graph:", COMPARISON_GRAPH_PATH)
    print("Heatmap:", HEATMAP_PATH)
    print("Result Graph:", RESULT_GRAPH_PATH)
    print("Prediction Graph:", PREDICTION_GRAPH_PATH)


if __name__ == "__main__":
    main()
```
