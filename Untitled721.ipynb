```python
import os
import json
import yaml
import joblib
import warnings
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

warnings.filterwarnings("ignore")

try:
    import tensorflow as tf
    from tensorflow.keras.models import Sequential
    from tensorflow.keras.layers import Dense, Dropout
except Exception:
    tf = None


BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Slum Risk Prediction System"
CSV_PATH = os.path.join(BASE_DIR, "D25-HousingandSlumPopulation_8_2.csv")

PKL_PATH = os.path.join(BASE_DIR, "slum_risk_model.pkl")
H5_PATH = os.path.join(BASE_DIR, "slum_risk_model.h5")
YAML_PATH = os.path.join(BASE_DIR, "slum_risk_config.yaml")
JSON_PATH = os.path.join(BASE_DIR, "slum_risk_metrics.json")

os.makedirs(BASE_DIR, exist_ok=True)


def clean_numeric(value):
    if pd.isna(value):
        return np.nan
    value = str(value)
    value = value.replace(",", "")
    value = value.replace("sex ratio", "")
    value = value.replace("%", "")
    value = value.strip()

    try:
        return float(value)
    except:
        return np.nan


def create_risk_label(df):
    """
    Risk is created using a rule-based score because the dataset
    does not already contain a target column.
    """

    df["Total Slums"] = (
        df["No. Notified Slums"]
        + df["No. of Recognised Slums"]
        + df["No. of Identfied Slums"]
    )

    df["Slum Density Score"] = df["Slum Population - Total"] / (df["Total Slums"] + 1)

    df["Vulnerability Score"] = (
        df["Slum Population - Total"] * 0.45
        + df["Population - SC Category"] * 0.20
        + df["Population - ST Category"] * 0.20
        + df["Slum Population - Child (0-6)"] * 0.15
    )

    if len(df) < 3:
        df["Risk Level"] = "High"
        return df

    q1 = df["Vulnerability Score"].quantile(0.33)
    q2 = df["Vulnerability Score"].quantile(0.66)

    def label(score):
        if score <= q1:
            return "Low"
        elif score <= q2:
            return "Medium"
        else:
            return "High"

    df["Risk Level"] = df["Vulnerability Score"].apply(label)
    return df


def main():
    print("Loading dataset...")
    df = pd.read_csv(CSV_PATH)

    print("Dataset shape:", df.shape)
    print("Columns:", list(df.columns))

    numeric_cols = [
        "Ward No.",
        "No. Notified Slums",
        "No. of Recognised Slums",
        "No. of Identfied Slums",
        "Slum Population - Total",
        "Slum Population - Male",
        "Slum Population - Female",
        "Slum Population - Child (0-6)",
        "Population - SC Category",
        "Population - ST Category",
        "Literacy Rate (%)",
    ]

    for col in numeric_cols:
        if col in df.columns:
            df[col] = df[col].apply(clean_numeric)

    df[numeric_cols] = df[numeric_cols].fillna(0)

    df = create_risk_label(df)

    feature_cols = [
        "No. Notified Slums",
        "No. of Recognised Slums",
        "No. of Identfied Slums",
        "Slum Population - Total",
        "Slum Population - Male",
        "Slum Population - Female",
        "Slum Population - Child (0-6)",
        "Population - SC Category",
        "Population - ST Category",
        "Literacy Rate (%)",
        "Total Slums",
        "Slum Density Score",
        "Vulnerability Score",
    ]

    X = df[feature_cols]
    y = df["Risk Level"]

    label_encoder = LabelEncoder()
    y_encoded = label_encoder.fit_transform(y)

    model_bundle = {
        "feature_columns": feature_cols,
        "label_encoder": label_encoder,
        "risk_classes": list(label_encoder.classes_),
    }

    if len(df) < 5 or len(set(y_encoded)) < 2:
        print("\nWARNING: Dataset is too small for proper ML training.")
        print("A rule-based model bundle will be saved as PKL.")
        print("Add more rows for real ML training.\n")

        model_bundle["model_type"] = "Rule-Based Risk Scoring"
        model_bundle["data"] = df
        joblib.dump(model_bundle, PKL_PATH)

        metrics = {
            "status": "rule_based_only",
            "reason": "Dataset has too few rows or only one class.",
            "rows": int(len(df)),
            "risk_classes": list(label_encoder.classes_),
            "output_files": {
                "pkl": PKL_PATH,
                "yaml": YAML_PATH,
                "json": JSON_PATH,
                "h5": "Not created because dataset is too small for neural training."
            }
        }

    else:
        X_train, X_test, y_train, y_test = train_test_split(
            X,
            y_encoded,
            test_size=0.2,
            random_state=42,
            stratify=y_encoded
        )

        sklearn_model = Pipeline([
            ("scaler", StandardScaler()),
            ("classifier", RandomForestClassifier(
                n_estimators=200,
                random_state=42,
                class_weight="balanced"
            ))
        ])

        print("Training Random Forest model...")
        sklearn_model.fit(X_train, y_train)

        y_pred = sklearn_model.predict(X_test)

        acc = accuracy_score(y_test, y_pred)

        model_bundle["model_type"] = "RandomForestClassifier"
        model_bundle["model"] = sklearn_model

        joblib.dump(model_bundle, PKL_PATH)

        metrics = {
            "status": "trained",
            "accuracy": float(acc),
            "classification_report": classification_report(
                y_test,
                y_pred,
                target_names=label_encoder.classes_,
                output_dict=True
            ),
            "confusion_matrix": confusion_matrix(y_test, y_pred).tolist(),
            "rows": int(len(df)),
            "features": feature_cols,
            "risk_classes": list(label_encoder.classes_),
            "output_files": {
                "pkl": PKL_PATH,
                "yaml": YAML_PATH,
                "json": JSON_PATH,
                "h5": H5_PATH
            }
        }

        if tf is not None:
            print("Training Neural Network model...")

            scaler = StandardScaler()
            X_train_scaled = scaler.fit_transform(X_train)
            X_test_scaled = scaler.transform(X_test)

            nn_model = Sequential([
                Dense(64, activation="relu", input_shape=(X_train_scaled.shape[1],)),
                Dropout(0.2),
                Dense(32, activation="relu"),
                Dense(len(label_encoder.classes_), activation="softmax")
            ])

            nn_model.compile(
                optimizer="adam",
                loss="sparse_categorical_crossentropy",
                metrics=["accuracy"]
            )

            nn_model.fit(
                X_train_scaled,
                y_train,
                epochs=50,
                batch_size=8,
                validation_data=(X_test_scaled, y_test),
                verbose=1
            )

            nn_model.save(H5_PATH)
        else:
            metrics["output_files"]["h5"] = "TensorFlow not installed. Run: pip install tensorflow"

    config = {
        "project_name": "Slum Risk Prediction System",
        "dataset_path": CSV_PATH,
        "base_directory": BASE_DIR,
        "target_column": "Risk Level",
        "feature_columns": feature_cols,
        "risk_logic": {
            "Low": "Lower vulnerability score",
            "Medium": "Moderate vulnerability score",
            "High": "Higher vulnerability score"
        },
        "artifacts": {
            "pkl_model": PKL_PATH,
            "h5_model": H5_PATH,
            "yaml_config": YAML_PATH,
            "json_metrics": JSON_PATH
        }
    }

    with open(YAML_PATH, "w") as f:
        yaml.dump(config, f, sort_keys=False)

    with open(JSON_PATH, "w") as f:
        json.dump(metrics, f, indent=4)

    print("\nArtifacts saved successfully:")
    print("PKL:", PKL_PATH)
    print("YAML:", YAML_PATH)
    print("JSON:", JSON_PATH)
    print("H5:", H5_PATH)


if __name__ == "__main__":
    main()
```
