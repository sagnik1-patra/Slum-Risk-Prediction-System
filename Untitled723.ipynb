```python
import os
import json
import yaml
import joblib
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.ensemble import RandomForestClassifier

warnings.filterwarnings("ignore")

try:
    import tensorflow as tf
    from tensorflow.keras.models import Sequential
    from tensorflow.keras.layers import Dense, Dropout
except Exception:
    tf = None


BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Slum Risk Prediction System"
CSV_PATH = os.path.join(BASE_DIR, "D25-HousingandSlumPopulation_8_2.csv")

os.makedirs(BASE_DIR, exist_ok=True)

PKL_PATH = os.path.join(BASE_DIR, "ais_slum_risk_model.pkl")
H5_PATH = os.path.join(BASE_DIR, "ais_slum_risk_model.h5")
YAML_PATH = os.path.join(BASE_DIR, "ais_slum_risk_config.yaml")
JSON_PATH = os.path.join(BASE_DIR, "ais_slum_risk_metrics.json")

RESULT_CSV_PATH = os.path.join(BASE_DIR, "ais_slum_risk_results.csv")
PREDICTION_CSV_PATH = os.path.join(BASE_DIR, "ais_slum_risk_predictions.csv")
COMPARISON_CSV_PATH = os.path.join(BASE_DIR, "ais_model_comparison_results.csv")

ACCURACY_GRAPH_PATH = os.path.join(BASE_DIR, "ais_accuracy_graph.png")
COMPARISON_GRAPH_PATH = os.path.join(BASE_DIR, "ais_comparison_graph.png")
HEATMAP_PATH = os.path.join(BASE_DIR, "ais_confusion_matrix_heatmap.png")
RESULT_GRAPH_PATH = os.path.join(BASE_DIR, "ais_result_distribution_graph.png")
PREDICTION_GRAPH_PATH = os.path.join(BASE_DIR, "ais_prediction_distribution_graph.png")
FEATURE_GRAPH_PATH = os.path.join(BASE_DIR, "ais_selected_feature_graph.png")


def clean_numeric(value):
    if pd.isna(value):
        return np.nan

    value = str(value)
    value = value.replace(",", "")
    value = value.replace("%", "")
    value = value.strip()

    try:
        return float(value)
    except:
        return np.nan


def prepare_dataset(df):
    numeric_cols = [
        "Ward No.",
        "No. Notified Slums",
        "No. of Recognised Slums",
        "No. of Identfied Slums",
        "Slum Population - Total",
        "Slum Population - Male",
        "Slum Population - Female",
        "Slum Population - Child (0-6)",
        "Population - SC Category",
        "Population - ST Category",
        "Literacy Rate (%)",
    ]

    for col in numeric_cols:
        if col not in df.columns:
            df[col] = 0
        df[col] = df[col].apply(clean_numeric)

    df[numeric_cols] = df[numeric_cols].fillna(0)

    df["Total Slums"] = (
        df["No. Notified Slums"]
        + df["No. of Recognised Slums"]
        + df["No. of Identfied Slums"]
    )

    df["Slum Density Score"] = (
        df["Slum Population - Total"] / (df["Total Slums"] + 1)
    )

    df["Gender Gap"] = abs(
        df["Slum Population - Male"] - df["Slum Population - Female"]
    )

    df["Child Population Ratio"] = (
        df["Slum Population - Child (0-6)"] /
        (df["Slum Population - Total"] + 1)
    )

    df["SC ST Population"] = (
        df["Population - SC Category"] + df["Population - ST Category"]
    )

    df["Vulnerability Score"] = (
        df["Slum Population - Total"] * 0.40
        + df["SC ST Population"] * 0.20
        + df["Slum Population - Child (0-6)"] * 0.20
        + df["Slum Density Score"] * 0.15
        - df["Literacy Rate (%)"] * 0.05
    )

    q1 = df["Vulnerability Score"].quantile(0.33)
    q2 = df["Vulnerability Score"].quantile(0.66)

    def risk_label(score):
        if score <= q1:
            return "Low"
        elif score <= q2:
            return "Medium"
        else:
            return "High"

    df["Risk Level"] = df["Vulnerability Score"].apply(risk_label)

    return df


def evaluate_feature_subset(X_train, X_test, y_train, y_test, feature_mask):
    selected_features = X_train.columns[feature_mask == 1]

    if len(selected_features) == 0:
        return 0

    model = Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", RandomForestClassifier(
            n_estimators=100,
            random_state=42,
            class_weight="balanced"
        ))
    ])

    model.fit(X_train[selected_features], y_train)
    preds = model.predict(X_test[selected_features])

    return accuracy_score(y_test, preds)


def ais_feature_selection(
    X_train,
    X_test,
    y_train,
    y_test,
    population_size=20,
    generations=25,
    mutation_rate=0.15
):
    n_features = X_train.shape[1]

    population = np.random.randint(0, 2, size=(population_size, n_features))

    best_antibody = None
    best_score = -1

    history = []

    for generation in range(generations):
        scores = []

        for antibody in population:
            score = evaluate_feature_subset(
                X_train,
                X_test,
                y_train,
                y_test,
                antibody
            )
            scores.append(score)

            if score > best_score:
                best_score = score
                best_antibody = antibody.copy()

        scores = np.array(scores)

        top_indices = scores.argsort()[-5:]
        top_antibodies = population[top_indices]

        clones = []

        for antibody in top_antibodies:
            for _ in range(4):
                clone = antibody.copy()

                for i in range(n_features):
                    if np.random.rand() < mutation_rate:
                        clone[i] = 1 - clone[i]

                if clone.sum() == 0:
                    clone[np.random.randint(0, n_features)] = 1

                clones.append(clone)

        random_antibodies = np.random.randint(
            0,
            2,
            size=(population_size - len(clones), n_features)
        )

        population = np.vstack([clones, random_antibodies])

        history.append({
            "Generation": generation + 1,
            "Best Accuracy": best_score,
            "Average Accuracy": float(np.mean(scores))
        })

        print(
            f"Generation {generation + 1}/{generations} | "
            f"Best Accuracy: {best_score:.4f} | "
            f"Average Accuracy: {np.mean(scores):.4f}"
        )

    selected_features = list(X_train.columns[best_antibody == 1])

    return selected_features, best_score, pd.DataFrame(history)


def save_count_graph(df, column, title, xlabel, ylabel, path):
    plt.figure(figsize=(8, 5))
    sns.countplot(data=df, x=column, order=df[column].value_counts().index)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.tight_layout()
    plt.savefig(path, dpi=300)
    plt.show()


def main():
    df = pd.read_csv(CSV_PATH)
    df = prepare_dataset(df)

    feature_cols = [
        "No. Notified Slums",
        "No. of Recognised Slums",
        "No. of Identfied Slums",
        "Slum Population - Total",
        "Slum Population - Male",
        "Slum Population - Female",
        "Slum Population - Child (0-6)",
        "Population - SC Category",
        "Population - ST Category",
        "Literacy Rate (%)",
        "Total Slums",
        "Slum Density Score",
        "Gender Gap",
        "Child Population Ratio",
        "SC ST Population",
        "Vulnerability Score",
    ]

    X = df[feature_cols]
    y = df["Risk Level"]

    label_encoder = LabelEncoder()
    y_encoded = label_encoder.fit_transform(y)

    if len(df) < 10 or len(set(y_encoded)) < 2:
        print("Dataset too small for AIS ML training. Saving rule-based AIS outputs.")

        df["Predicted Risk Level"] = df["Risk Level"]
        df["Prediction Status"] = "AIS Rule-Based"

        df.to_csv(RESULT_CSV_PATH, index=False)
        df.to_csv(PREDICTION_CSV_PATH, index=False)

        save_count_graph(
            df,
            "Risk Level",
            "AIS Result Risk Distribution",
            "Risk Level",
            "Count",
            RESULT_GRAPH_PATH
        )

        save_count_graph(
            df,
            "Predicted Risk Level",
            "AIS Prediction Risk Distribution",
            "Predicted Risk Level",
            "Count",
            PREDICTION_GRAPH_PATH
        )

        joblib.dump({
            "model_type": "AIS Rule-Based",
            "feature_columns": feature_cols,
            "label_encoder": label_encoder
        }, PKL_PATH)

        config = {
            "project_name": "AIS Slum Risk Prediction System",
            "status": "rule_based_only",
            "dataset_path": CSV_PATH,
            "outputs": {
                "pkl": PKL_PATH,
                "result_csv": RESULT_CSV_PATH,
                "prediction_csv": PREDICTION_CSV_PATH,
                "result_graph": RESULT_GRAPH_PATH,
                "prediction_graph": PREDICTION_GRAPH_PATH
            }
        }

        metrics = {
            "status": "rule_based_only",
            "rows": int(len(df)),
            "message": "Add more rows for AIS training accuracy, heatmap, and comparison graph."
        }

        with open(YAML_PATH, "w") as f:
            yaml.dump(config, f, sort_keys=False)

        with open(JSON_PATH, "w") as f:
            json.dump(metrics, f, indent=4)

        print("AIS rule-based files saved successfully.")
        return

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y_encoded,
        test_size=0.2,
        random_state=42,
        stratify=y_encoded
    )

    print("Running AIS feature selection...")

    selected_features, ais_best_score, ais_history = ais_feature_selection(
        X_train,
        X_test,
        y_train,
        y_test
    )

    ais_history.to_csv(COMPARISON_CSV_PATH, index=False)

    print("AIS selected features:", selected_features)

    final_model = Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", RandomForestClassifier(
            n_estimators=300,
            random_state=42,
            class_weight="balanced"
        ))
    ])

    final_model.fit(X_train[selected_features], y_train)

    y_pred = final_model.predict(X_test[selected_features])
    accuracy = accuracy_score(y_test, y_pred)

    df["Predicted Risk Encoded"] = final_model.predict(X[selected_features])
    df["Predicted Risk Level"] = label_encoder.inverse_transform(
        df["Predicted Risk Encoded"]
    )

    df["Prediction Status"] = np.where(
        df["Risk Level"] == df["Predicted Risk Level"],
        "Correct",
        "Incorrect"
    )

    df.to_csv(RESULT_CSV_PATH, index=False)

    prediction_df = df[
        [
            "Risk Level",
            "Predicted Risk Level",
            "Prediction Status",
            "Vulnerability Score",
            "Slum Density Score",
            "Total Slums"
        ]
    ]

    prediction_df.to_csv(PREDICTION_CSV_PATH, index=False)

    joblib.dump({
        "model_type": "AIS + Random Forest",
        "model": final_model,
        "selected_features": selected_features,
        "all_features": feature_cols,
        "label_encoder": label_encoder,
        "risk_classes": list(label_encoder.classes_)
    }, PKL_PATH)

    metrics = {
        "status": "trained",
        "algorithm": "Artificial Immune System Feature Selection",
        "final_model": "Random Forest",
        "accuracy": float(accuracy),
        "ais_best_feature_selection_score": float(ais_best_score),
        "selected_features": selected_features,
        "classification_report": classification_report(
            y_test,
            y_pred,
            target_names=label_encoder.classes_,
            output_dict=True
        ),
        "confusion_matrix": confusion_matrix(y_test, y_pred).tolist()
    }

    with open(JSON_PATH, "w") as f:
        json.dump(metrics, f, indent=4)

    config = {
        "project_name": "AIS Slum Risk Prediction System",
        "dataset_path": CSV_PATH,
        "target_column": "Risk Level",
        "algorithm": "AIS Feature Selection",
        "final_model": "Random Forest",
        "selected_features": selected_features,
        "artifacts": {
            "pkl": PKL_PATH,
            "h5": H5_PATH,
            "yaml": YAML_PATH,
            "json": JSON_PATH,
            "result_csv": RESULT_CSV_PATH,
            "prediction_csv": PREDICTION_CSV_PATH,
            "accuracy_graph": ACCURACY_GRAPH_PATH,
            "comparison_graph": COMPARISON_GRAPH_PATH,
            "heatmap": HEATMAP_PATH,
            "result_graph": RESULT_GRAPH_PATH,
            "prediction_graph": PREDICTION_GRAPH_PATH
        }
    }

    with open(YAML_PATH, "w") as f:
        yaml.dump(config, f, sort_keys=False)

    plt.figure(figsize=(10, 6))
    plt.plot(ais_history["Generation"], ais_history["Best Accuracy"], marker="o", label="Best Accuracy")
    plt.plot(ais_history["Generation"], ais_history["Average Accuracy"], marker="o", label="Average Accuracy")
    plt.title("AIS Accuracy Graph")
    plt.xlabel("Generation")
    plt.ylabel("Accuracy")
    plt.legend()
    plt.tight_layout()
    plt.savefig(ACCURACY_GRAPH_PATH, dpi=300)
    plt.show()

    plt.figure(figsize=(10, 6))
    sns.barplot(data=ais_history, x="Generation", y="Best Accuracy")
    plt.title("AIS Model Comparison by Generation")
    plt.xlabel("AIS Generation")
    plt.ylabel("Best Accuracy")
    plt.tight_layout()
    plt.savefig(COMPARISON_GRAPH_PATH, dpi=300)
    plt.show()

    cm = confusion_matrix(y_test, y_pred)

    plt.figure(figsize=(8, 6))
    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=label_encoder.classes_,
        yticklabels=label_encoder.classes_
    )
    plt.title("AIS Confusion Matrix Heatmap")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.tight_layout()
    plt.savefig(HEATMAP_PATH, dpi=300)
    plt.show()

    save_count_graph(
        df,
        "Risk Level",
        "AIS Actual Result Distribution",
        "Actual Risk Level",
        "Count",
        RESULT_GRAPH_PATH
    )

    save_count_graph(
        df,
        "Predicted Risk Level",
        "AIS Prediction Distribution",
        "Predicted Risk Level",
        "Count",
        PREDICTION_GRAPH_PATH
    )

    selected_feature_df = pd.DataFrame({
        "Feature": selected_features,
        "Selected": [1] * len(selected_features)
    })

    plt.figure(figsize=(10, 6))
    sns.barplot(data=selected_feature_df, x="Selected", y="Feature")
    plt.title("AIS Selected Features")
    plt.xlabel("Selected")
    plt.ylabel("Feature")
    plt.tight_layout()
    plt.savefig(FEATURE_GRAPH_PATH, dpi=300)
    plt.show()

    if tf is not None:
        scaler = StandardScaler()
        X_train_scaled = scaler.fit_transform(X_train[selected_features])
        X_test_scaled = scaler.transform(X_test[selected_features])

        nn_model = Sequential([
            Dense(64, activation="relu", input_shape=(X_train_scaled.shape[1],)),
            Dropout(0.2),
            Dense(32, activation="relu"),
            Dropout(0.2),
            Dense(len(label_encoder.classes_), activation="softmax")
        ])

        nn_model.compile(
            optimizer="adam",
            loss="sparse_categorical_crossentropy",
            metrics=["accuracy"]
        )

        nn_model.fit(
            X_train_scaled,
            y_train,
            validation_data=(X_test_scaled, y_test),
            epochs=50,
            batch_size=8,
            verbose=1
        )

        nn_model.save(H5_PATH)

    print("\nAIS files saved successfully:")
    print(PKL_PATH)
    print(H5_PATH)
    print(YAML_PATH)
    print(JSON_PATH)
    print(RESULT_CSV_PATH)
    print(PREDICTION_CSV_PATH)
    print(ACCURACY_GRAPH_PATH)
    print(COMPARISON_GRAPH_PATH)
    print(HEATMAP_PATH)
    print(RESULT_GRAPH_PATH)
    print(PREDICTION_GRAPH_PATH)
    print(FEATURE_GRAPH_PATH)


if __name__ == "__main__":
    main()
```
